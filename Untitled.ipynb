```python
import whisper
import subprocess
import os

VIDEO_PATH = ""
WAV_PATH = "output_audio.wav"

def extract_audio(video_path, wav_path, sample_rate=16000):
    """
    Extract clean mono WAV audio from video using ffmpeg
    """
    command = [
        "ffmpeg",
        "-y",
        "-i", video_path,
        "-vn",
        "-ac", "1",
        "-ar", str(sample_rate),
        "-acodec", "pcm_s16le",
        wav_path
    ]
    subprocess.run(command, check=True)

def transcribe_audio(wav_path, model_size="base"):
    """
    Transcribe audio using Whisper Python library
    """
    model = whisper.load_model(model_size)
    result = model.transcribe(wav_path)
    return result["text"]

if __name__ == "__main__":
    extract_audio(VIDEO_PATH, WAV_PATH)
    print(f"Audio saved to {WAV_PATH}")

    transcription = transcribe_audio(WAV_PATH, model_size="base")
    print("Transcription:")
    print(transcription)
```
